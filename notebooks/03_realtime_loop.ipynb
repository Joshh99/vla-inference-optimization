# 20 Hz inference-loop sketch

This notebook shows the control-loop boundary around the reusable runner. Replace `read_episode()` and `send_action()` with robot or simulator adapters.

```python
import time
from vla_inference import SmolVLARunner

runner = SmolVLARunner()
runner.initialize('lerobot/smolvla_base')
```

```python
period_s = 1 / 20
runner.reset()
for _ in range(100):
    started = time.perf_counter()
    episode = read_episode()
    action_chunk = runner.predict(episode)
    send_action(action_chunk[0])
    time.sleep(max(0, period_s - (time.perf_counter() - started)))
```
